```python
# identify language rows
lang_specific_state_df["language"] = lang_specific_state_df["Label (Grouping)"].where(
    lang_specific_state_df["Label (Grouping)"].str.endswith(":")
)
```

```python
# carry language down
lang_specific_state_df["language"] = lang_specific_state_df["language"].ffill(limit=2).str.strip(":").str.strip()
```

```python
lang_specific_state_df_copy=lang_specific_state_df.copy()
```

```python
lang_specific_state_df_copy=lang_specific_state_df_copy.rename(columns=
                                                                     {'Label (Grouping)': 'proficiency',
                                                                     'Maryland!!Estimate': 'estimated_speakers_count',
                                                                     'Maryland!!Margin of Error': 'moe'})
```

```python
lang_specific_state_df_with_total=lang_specific_state_df_copy.copy()
```

```python
lang_specific_state_df_copy.at[1,'language']='English'
```

```python
lang_specific_state_df_copy=lang_specific_state_df_copy.drop(0)
```

```python
lang_specific_state_df_copy['estimated_speakers_count']=lang_specific_state_df_copy['estimated_speakers_count'].str.replace(",","")
```

```python
lang_specific_state_df_copy['moe']=lang_specific_state_df_copy['moe'].str.replace(",","")
```

```python
lang_specific_state_df_copy['proficiency'] = lang_specific_state_df_copy['proficiency'].str.strip(":").str.strip()
```

```python
lang_specific_state_df_copy.head()
```

```python
lang_specific_state_df_copy['estimated_speakers_count']=lang_specific_state_df_copy['estimated_speakers_count'].astype(int)
```

```python
# lang_specific_state_df_with_total
# 5912247
```

```python
lang_specific_state_df_copy['pct_of_total_pop'] = ((lang_specific_state_df_copy['estimated_speakers_count']/5912247)*100).round(2)
```

```python
lang_specific_proficiency_state_df = lang_specific_state_df_copy.copy()
```

```python
lang_specific_proficiency_state_df.head()
```

```python
lang_specific_proficiency_state_df['speaker_count_by_language'] = (lang_specific_proficiency_state_df['estimated_speakers_count']
    .where(~lang_specific_proficiency_state_df['proficiency'].str.contains("Speak English", na=False))).ffill(limit=2)
```

```python
lang_specific_proficiency_state_df.head()
```

```python
lang_specific_proficiency_state_df['proficiency_pct_by_language'] = ((lang_specific_proficiency_state_df['estimated_speakers_count']/lang_specific_proficiency_state_df['speaker_count_by_language'])*100).round(2)
# lang_specific_proficiency_state_df['proficiency_pct_by_language'] = (lang_specific_proficiency_state_df['proficiency_pct_by_language']
#     .where(lang_specific_proficiency_state_df['proficiency_pct_by_language'] != 100.00))
```

```python
lang_specific_proficiency_state_df['proficiency'] = (lang_specific_proficiency_state_df['proficiency'].where(lang_specific_proficiency_state_df['proficiency'].str.contains("Speak English|Speak only English", na=False))
)
```

```python
lang_specific_proficiency_state_df
```

```python
stacked_specific_lang_proficiency_state_df = (lang_specific_proficiency_state_df.pivot_table(index='language',
                                                                                             columns='proficiency',
                                                                                             values='proficiency_pct_by_language',
                                                                                             aggfunc='sum').fillna(0).reset_index())
```

```python
stacked_specific_lang_proficiency_state_df

speaker_totals = (lang_specific_proficiency_state_df[['language', 'speaker_count_by_language']].drop_duplicates())

stacked_specific_lang_proficiency_state_df = (stacked_specific_lang_proficiency_state_df
    .merge(speaker_totals, on='language', how='left')
```

```python
stacked_specific_lang_proficiency_state_df['speaker_count_by_language']=stacked_specific_lang_proficiency_state_df['speaker_count_by_language'].astype(int)
```

```python
stacked_specific_lang_proficiency_state_df=stacked_specific_lang_proficiency_state_df.drop('Speak only English', axis=1)
```
