```python
#language by percentage
languages_pct_df_long = languages_md_county_df_long.melt(
    id_vars = ['Geography', 'Geographic Area Name'],
    value_vars = ['spanish_pct_65_plus',
                  
                  'indo_euro_pct_65_plus',
                  
                  'asian_pacific_pct_65_plus',
                
                  'othamer_lang_pct_65_plus'],
    var_name = 'language',
    value_name = 'pct_65_plus')
    
#language spoken at home among LEP by percentage
languages_lep_pct_df_long = languages_md_county_df_long.melt(
    id_vars = ['Geography', 'Geographic Area Name'],
    value_vars = ['lep_pct_over_5',
                
                  'lep_pct_spanish', 
            
                  'lep_pct_spanish_65_plus',
                 
                  'lep_pct_indo_euro',
                
                  'lep_pct_indo_euro_65_plus',
                 
                  'lep_pct_asian_pacific',
              
                  'lep_pct_asian_pacific_65_plus',
                  
                  'lep_pct_other',
              
                  'lep_pct_other_65_plus'],
    var_name = 'language',
    value_name = 'lep_pct')
```

```python
#MOE language by percentage
moe_languages_pct_df_long = languages_md_county_df_long.melt(
    id_vars = ['Geography', 'Geographic Area Name'],
    value_vars = [
                  'moe_spanish_pct_65_plus',
                  
                  'moe_indo_euro_pct_65_plus',
                  
                  'moe_asian_pacific_pct_65_plus',
                  
                 'moe_other_lang_pct_65_plus'],
    var_name = 'language',
    value_name = 'moe_pct_65_plus')

    
#MOE language spoken at home among LEP by percentage
moe_languages_lep_pct_df_long = languages_md_county_df_long.melt(
    id_vars = ['Geography', 'Geographic Area Name'],
    value_vars = [
                  'moe_lep_pct_over_5',
                  
                  'moe_lep_pct_spanish',
       
                  'moe_lep_pct_spanish_65_plus',
            
                  'moe_lep_pct_indo_euro',
                
                  'moe_lep_pct_indo_euro_65_plus',
              
                  'moe_lep_pct_asian_pacific',
               
                  'moe_lep_pct_asian_pacific_65_plus',
          
                  'moe_lep_pct_other',
              
                  'moe_lep_pct_other_65_plus'],
    var_name = 'language',
    value_name = 'moe_lep_pct_65_plus')
```

```python
moe_languages_pct_df_long['language'] = moe_languages_pct_df_long['language'].str.replace('moe_','')
moe_languages_lep_pct_df_long['language'] = moe_languages_lep_pct_df_long['language'].str.replace('moe_','')
```

```python
languages_pct_moe_df_long = pd.merge(languages_pct_df_long, moe_languages_pct_df_long,
                                       on=['Geography', 'Geographic Area Name', 'language'], 
                                       how='inner')

languages_lep_pct_moe_df_long = pd.merge(languages_lep_pct_df_long, moe_languages_lep_pct_df_long,
                                       on=['Geography', 'Geographic Area Name', 'language'], 
                                       how='inner')
```

```python
LEP_counts_total_65_plus = languages_md_county_df[[
                                                   'Estimate!!Speak English  less than very well"!!Percent of specified language speakers!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Spanish!!65 years old and over"',
                                                   'Estimate!!Speak English  less than very well"!!Percent of specified language speakers!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Other Indo-European languages!!65 years old and over"',
                                                   'Estimate!!Speak English  less than very well"!!Percent of specified language speakers!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Asian and Pacific Island languages!!65 years old and over"',
                                                   'Estimate!!Speak English  less than very well"!!Percent of specified language speakers!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Other languages!!65 years old and over"']]
LEP_counts_total_65_plus=LEP_counts_total_65_plus.astype(int)
languages_md_county_df["LEP_counts_total_65_plus"] = LEP_counts_total_65_plus.sum(axis=1)
```

```python
language_speakers_counts_total_65_plus = languages_md_county_df[[
'Estimate!!Total!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Spanish!!65 years old and over',
'Estimate!!Total!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Other Indo-European languages!!65 years old and over',
'Estimate!!Total!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Asian and Pacific Island languages!!65 years old and over',
'Estimate!!Total!!Population 5 years and over!!SPEAK A LANGUAGE OTHER THAN ENGLISH!!Other languages!!65 years old and over']]

language_speakers_counts_total_65_plus=language_speakers_counts_total_65_plus.astype(int)

languages_md_county_df["language_speakers_counts_total_65_plus"] = (language_speakers_counts_total_65_plus.sum(axis=1))
```

```python
languages_md_county_df['pct_LEP_language_speakers'] = (languages_md_county_df["LEP_counts_total_65_plus"]/languages_md_county_df["language_speakers_counts_total_65_plus"])*100
```

```python
languages_lep_pct_moe_df_long_wide=languages_lep_pct_moe_df_long_wide.merge(languages_md_county_df[['Geographic Area Name', 'pct_LEP_language_speakers',"language_speakers_counts_total_65_plus", 'LEP_counts_total_65_plus']],
        on='Geographic Area Name',
        how='left').round(2)
```

```python
languages_lep_pct_moe_df_long_wide['pct_language_speakers'] = (languages_lep_pct_moe_df_long_wide["language_speakers_counts_total_65_plus"]/languages_lep_pct_moe_df_long_wide["total_65_count"])*100
```

```python
languages_lep_pct_moe_df_long_wide['pct_language_speakers']=languages_lep_pct_moe_df_long_wide['pct_language_speakers'].round(2)
```
